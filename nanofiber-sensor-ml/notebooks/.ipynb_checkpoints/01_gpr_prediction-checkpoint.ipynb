# Project 1 — Gaussian Process Regression for Pressure Sensor Sensitivity

**Goal:** Fit a GPR model to the sensitivity-vs-pressure curves of two sensor types
(coaxial hollow fiber and single control fiber), predict the full curve with
uncertainty bands, and evaluate model performance.

**Data:** `Sensitivity_Comparison.csv` — 3 replicates per sensor type across
33 pressure points (0.2 – 300 kPa).

**Paper:** Siddique et al., *IEEE Journal on Flexible Electronics*, 2025.
DOI: 10.1109/JFLEX.2025.3577111

---
## Why GPR for this data?

| Method | Requires | Your dataset |
|--------|----------|--------------|
| Neural network | 1000s of samples | 33 pressure points |
| Random forest | ~100+ samples | 33 pressure points |
| **GPR** | **~10–100 samples** | **33 pressure points ✓** |

GPR is the standard choice in materials informatics for small experimental datasets.
It gives **uncertainty estimates automatically** — no extra code needed.


````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec
from scipy import stats

from sklearn.gaussian_process import GaussianProcessRegressor
from sklearn.gaussian_process.kernels import (
    RBF, Matern, WhiteKernel, ConstantKernel as C
)
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import LeaveOneOut
from sklearn.metrics import mean_absolute_error, r2_score

import warnings
warnings.filterwarnings('ignore')

print("All imports successful.")
print(f"scikit-learn version: ", end="")
import sklearn; print(sklearn.__version__)
````

## 1. Load and inspect data

````python
# ── Load CSV (3-row header, '--' as missing) ───────────────────────────────
raw = pd.read_csv(
    r'C:\Users\hp\Downloads\nanofiber-sensor-ml\nanofiber-sensor-ml\data\Sensitivity_Comparison.csv',
    header=None,
    skiprows=3,
    na_values=['--', '']
)

raw.columns = [
    'pressure',
    'ctrl_1', 'ctrl_2', 'ctrl_3',
    'ctrl_mean_orig', 'ctrl_sd_orig',
    'blank',
    'coax_1', 'coax_2', 'coax_3',
    'coax_hex1', 'coax_hex5',
    'coax_hex7', 'coax_hex8', 'coax_hex9',
    'coax_mean_orig', 'coax_sd_orig'
]

# Convert all numeric columns
for col in raw.columns:
    raw[col] = pd.to_numeric(raw[col], errors='coerce')

df = raw.dropna(subset=['pressure']).copy()

# ── Compute mean, SD, SEM from the 3 main replicates ──────────────────────
n = 3
t_crit = stats.t.ppf(0.975, df=n-1)  # 4.303 for n=3

for sensor, cols in [('ctrl', ['ctrl_1','ctrl_2','ctrl_3']),
                     ('coax', ['coax_1','coax_2','coax_3'])]:
    df[f'{sensor}_mean'] = df[cols].mean(axis=1)
    df[f'{sensor}_sd']   = df[cols].std(axis=1, ddof=1)
    df[f'{sensor}_sem']  = df[f'{sensor}_sd'] / np.sqrt(n)
    df[f'{sensor}_ci95'] = t_crit * df[f'{sensor}_sem']

print(f"Loaded {len(df)} pressure points")
print(f"Pressure range: {df['pressure'].min()} – {df['pressure'].max()} kPa")
print(f"\nFirst 5 rows (key columns):")
df[['pressure','ctrl_mean','ctrl_sd','coax_mean','coax_sd']].head()
````

````python
# ── Quick statistical summary ─────────────────────────────────────────────
print("Control sensor — summary statistics:")
print(df['ctrl_mean'].describe().round(4))
print("\nCoaxial sensor — summary statistics:")
print(df['coax_mean'].describe().round(4))

# Missing data check
missing = df[['coax_1','coax_2','coax_3','ctrl_1','ctrl_2','ctrl_3']].isna().sum()
print("\nMissing values per column:")
print(missing[missing > 0] if missing.sum() > 0 else "None — dataset is complete")
````

## 2. Feature engineering — log-transform pressure

The pressure axis spans 3 orders of magnitude (0.2 → 300 kPa).
GPR kernels compute distances in input space. Without log-transform,
the model would treat the gap between 0.2 and 1 kPa as negligible
compared to 100–300 kPa — the opposite of what matters physically.

`log10(P)` maps the range to [-0.7, 2.5], giving equal distance weight
across the full range.


````python
# ── Log-transform pressure (essential for multi-decade x-axis) ────────────
df['log_pressure'] = np.log10(df['pressure'])

# Build X (features) and y arrays for each sensor type
# Shape: (n_points, 1) — single feature (log pressure)
X = df['log_pressure'].values.reshape(-1, 1)

y_ctrl = df['ctrl_mean'].values
y_coax = df['coax_mean'].values

# Noise level estimates from standard deviation
# GPR uses this as the measurement noise (alpha parameter)
noise_ctrl = df['ctrl_sd'].fillna(df['ctrl_sd'].median()).values
noise_coax = df['coax_sd'].fillna(df['coax_sd'].median()).values

print(f"X shape : {X.shape}   (log10 pressure, {len(X)} points)")
print(f"y_ctrl  : min={y_ctrl.min():.3f}  max={y_ctrl.max():.3f}  mean={y_ctrl.mean():.3f}")
print(f"y_coax  : min={y_coax.min():.3f}  max={y_coax.max():.3f}  mean={y_coax.mean():.3f}")
print(f"\nLog-pressure range: [{X.min():.2f}, {X.max():.2f}]")

# Dense prediction grid (200 points) for smooth curves
X_pred_log = np.linspace(X.min() - 0.05, X.max() + 0.05, 200).reshape(-1, 1)
X_pred_kpa = 10 ** X_pred_log.ravel()  # back to kPa for plotting
````

## 3. Kernel selection — the most important GPR decision

The kernel defines what "similarity" means between pressure points.
We use a **Matérn ν=2.5 kernel** — the standard choice for physical
processes because it assumes the function is twice differentiable
(smooth but not infinitely smooth like RBF).

```
k(x, x') = C × Matérn(ν=2.5, length_scale=l) + WhiteKernel(noise)
│           │                                   └─ models measurement noise
│           └─ Matérn: physical smoothness          
└─ C: amplitude (scales with sensitivity range)
```

The length scale `l` controls how far apart two pressure points can be
and still influence each other. GPR **optimises all kernel parameters
automatically** by maximising the log-marginal-likelihood on your data.


````python
# ── Define kernel ─────────────────────────────────────────────────────────
# C: amplitude scale  (init=1, bounds allow 0.01–100)
# Matern nu=2.5: physical smoothness
# WhiteKernel: heteroscedastic noise floor
kernel = (
    C(1.0, constant_value_bounds=(0.01, 100))
    * Matern(length_scale=0.5, nu=2.5,
             length_scale_bounds=(0.01, 10))
    + WhiteKernel(noise_level=0.01,
                  noise_level_bounds=(1e-5, 1))
)

print("Kernel definition:")
print(kernel)
print("\nKernel parameters will be optimised automatically during fit.")
````

## 4. Fit GPR models

````python
# ── Fit GPR for CONTROL sensor ────────────────────────────────────────────
gpr_ctrl = GaussianProcessRegressor(
    kernel=kernel,
    alpha=noise_ctrl**2,   # per-point noise variance from your SD data
    n_restarts_optimizer=10,
    normalize_y=True,
    random_state=42
)
gpr_ctrl.fit(X, y_ctrl)

# ── Fit GPR for COAXIAL sensor ────────────────────────────────────────────
gpr_coax = GaussianProcessRegressor(
    kernel=kernel,
    alpha=noise_coax**2,
    n_restarts_optimizer=10,
    normalize_y=True,
    random_state=42
)
gpr_coax.fit(X, y_coax)

print("Control sensor — optimised kernel:")
print(" ", gpr_ctrl.kernel_)
print(f"  Log-marginal-likelihood: {gpr_ctrl.log_marginal_likelihood_value_:.3f}")

print("\nCoaxial sensor — optimised kernel:")
print(" ", gpr_coax.kernel_)
print(f"  Log-marginal-likelihood: {gpr_coax.log_marginal_likelihood_value_:.3f}")
````

````python
# ── Predict over dense grid ───────────────────────────────────────────────
mu_ctrl, std_ctrl = gpr_ctrl.predict(X_pred_log, return_std=True)
mu_coax, std_coax = gpr_coax.predict(X_pred_log, return_std=True)

# Clip to physically meaningful range (sensitivity >= 0)
mu_ctrl = np.clip(mu_ctrl, 0, None)
mu_coax = np.clip(mu_coax, 0, None)

# 95% CI = mean ± 1.96 × std  (GPR std is already the posterior std)
ci_ctrl = 1.96 * std_ctrl
ci_coax = 1.96 * std_coax

print("Prediction complete.")
print(f"Control  — peak predicted: {mu_ctrl.max():.4f} kPa⁻¹ @ {X_pred_kpa[mu_ctrl.argmax()]:.2f} kPa")
print(f"Coaxial  — peak predicted: {mu_coax.max():.4f} kPa⁻¹ @ {X_pred_kpa[mu_coax.argmax()]:.2f} kPa")

# Crossover point
diff = mu_coax - mu_ctrl
sign_changes = np.where(np.diff(np.sign(diff)))[0]
if len(sign_changes) > 0:
    cross_kpa = X_pred_kpa[sign_changes[0]]
    print(f"Crossover pressure (GPR): ~{cross_kpa:.1f} kPa")
````

````python
# ── Figure 1 : GPR fit with uncertainty bands ─────────────────────────────
COLOR_COAX = '#1D9E75'
COLOR_CTRL = '#7F77DD'

fig, axes = plt.subplots(1, 2, figsize=(14, 5.5))

for ax, xscale, title, xlim in [
    (axes[0], 'linear', 'Full pressure range (linear)', (0, 305)),
    (axes[1], 'log',    'Low-pressure regime (log scale)', (0.15, 12))
]:
    # ── Coaxial ──
    ax.plot(X_pred_kpa, mu_coax, color=COLOR_COAX, lw=2,
            label='Coaxial hollow fiber — GPR mean')
    ax.fill_between(X_pred_kpa,
                    np.clip(mu_coax - ci_coax, 0, None),
                    mu_coax + ci_coax,
                    alpha=0.18, color=COLOR_COAX, label='95% CI')
    ax.errorbar(df['pressure'], df['coax_mean'], yerr=df['coax_sd'],
                fmt='o', color=COLOR_COAX, ms=4.5, capsize=3,
                lw=0.8, zorder=4, label='Measured (n=3)')

    # ── Control ──
    ax.plot(X_pred_kpa, mu_ctrl, color=COLOR_CTRL, lw=2, ls='--',
            label='Single control fiber — GPR mean')
    ax.fill_between(X_pred_kpa,
                    np.clip(mu_ctrl - ci_ctrl, 0, None),
                    mu_ctrl + ci_ctrl,
                    alpha=0.18, color=COLOR_CTRL, label='95% CI')
    ax.errorbar(df['pressure'], df['ctrl_mean'], yerr=df['ctrl_sd'],
                fmt='s', color=COLOR_CTRL, ms=4.5, capsize=3,
                lw=0.8, zorder=4, label='Measured (n=3)')

    # Crossover line
    if len(sign_changes) > 0:
        ax.axvline(cross_kpa, color='#D85A30', ls=':', lw=1.2, alpha=0.7)

    ax.set_xscale(xscale)
    ax.set_xlim(xlim)
    ax.set_ylim(-0.02, None)
    ax.set_xlabel('Pressure (kPa)', fontsize=11)
    ax.set_ylabel('Sensitivity (kPa⁻¹)', fontsize=11)
    ax.set_title(title, fontsize=11)
    ax.grid(True, alpha=0.2)

# Single combined legend on left panel
handles, labels = axes[1].get_legend_handles_labels()
# Deduplicate
seen, h2, l2 = set(), [], []
for h, l in zip(handles, labels):
    if l not in seen:
        seen.add(l); h2.append(h); l2.append(l)
axes[0].legend(h2, l2, fontsize=8.5, loc='upper right')

fig.suptitle(
    'GPR fit — Sensitivity vs Pressure\nCoaxial hollow fiber vs Single control fiber  '
    '(Siddique et al., IEEE JFLEX 2025)',
    fontsize=11, y=1.01
)
plt.tight_layout()
plt.savefig('outputs/01_gpr_fit.png', dpi=200, bbox_inches='tight')
plt.show()
print("Saved: outputs/01_gpr_fit.png")
````

## 6. Leave-One-Out Cross-Validation

With only 33 data points, LOO-CV is the correct accuracy metric.
Each iteration removes one pressure point, re-trains on the rest,
and predicts the removed point. This gives an honest measure of
how well GPR generalises to **unseen pressures**.


````python
# ── LOO-CV for both sensors ───────────────────────────────────────────────
def loo_cv(X, y, noise, kernel, label):
    loo = LeaveOneOut()
    preds, actuals = [], []
    for train_idx, test_idx in loo.split(X):
        gpr_tmp = GaussianProcessRegressor(
            kernel=kernel,
            alpha=noise[train_idx]**2,
            n_restarts_optimizer=3,
            normalize_y=True,
            random_state=42
        )
        gpr_tmp.fit(X[train_idx], y[train_idx])
        pred = gpr_tmp.predict(X[test_idx])[0]
        preds.append(max(pred, 0))   # clip negative predictions
        actuals.append(y[test_idx][0])
    
    preds   = np.array(preds)
    actuals = np.array(actuals)
    mae     = mean_absolute_error(actuals, preds)
    r2      = r2_score(actuals, preds)
    mape    = np.mean(np.abs((actuals - preds) / (actuals + 1e-9))) * 100
    
    print(f"\n{label}")
    print(f"  MAE  : {mae:.4f} kPa⁻¹  (mean absolute error)")
    print(f"  R²   : {r2:.4f}         (1.0 = perfect)")
    print(f"  MAPE : {mape:.2f}%      (mean absolute % error)")
    return preds, actuals, mae, r2

preds_ctrl, actuals_ctrl, mae_ctrl, r2_ctrl = loo_cv(
    X, y_ctrl, noise_ctrl, kernel, "Control sensor (LOO-CV)"
)
preds_coax, actuals_coax, mae_coax, r2_coax = loo_cv(
    X, y_coax, noise_coax, kernel, "Coaxial sensor (LOO-CV)"
)
````

````python
# ── Figure 2 : LOO-CV diagnostic — predicted vs actual ────────────────────
fig, axes = plt.subplots(1, 2, figsize=(11, 5))

for ax, preds, actuals, mae, r2, color, label in [
    (axes[0], preds_ctrl, actuals_ctrl, mae_ctrl, r2_ctrl,
     COLOR_CTRL, 'Control sensor'),
    (axes[1], preds_coax, actuals_coax, mae_coax, r2_coax,
     COLOR_COAX, 'Coaxial sensor'),
]:
    lims = [min(actuals.min(), preds.min()) * 0.9,
            max(actuals.max(), preds.max()) * 1.1]
    ax.plot(lims, lims, 'k--', lw=0.8, alpha=0.4, label='Perfect prediction')
    ax.scatter(actuals, preds, color=color, alpha=0.8, s=40, zorder=3)
    ax.set_xlabel('Measured sensitivity (kPa⁻¹)', fontsize=11)
    ax.set_ylabel('GPR predicted sensitivity (kPa⁻¹)', fontsize=11)
    ax.set_title(f'{label}\nMAE = {mae:.4f}  |  R² = {r2:.4f}', fontsize=11)
    ax.set_xlim(lims); ax.set_ylim(lims)
    ax.grid(True, alpha=0.2)
    ax.text(0.05, 0.92, f'R² = {r2:.3f}', transform=ax.transAxes,
            fontsize=10, color=color, fontweight='bold')

plt.suptitle('LOO Cross-Validation: GPR Predicted vs Measured Sensitivity', fontsize=11)
plt.tight_layout()
plt.savefig('outputs/01_gpr_loo_cv.png', dpi=200, bbox_inches='tight')
plt.show()
print("Saved: outputs/01_gpr_loo_cv.png")
````

## 7. Uncertainty analysis — where does the model know less?

GPR uncertainty is not uniform. It is larger:
- **Between measured pressure points** (no nearby data)
- **At the tails** (extrapolation beyond measurement range)
- **Where your sensors disagreed** (high SD → model given noisy signal)

This plot is directly useful for experiment planning: the high-uncertainty
regions tell you where to run additional experiments.


````python
# ── Figure 3 : Uncertainty heatmap ────────────────────────────────────────
fig, axes = plt.subplots(2, 1, figsize=(12, 7), sharex=True)

for ax, mu, std, y_data, noise, color, label in [
    (axes[0], mu_ctrl, std_ctrl, y_ctrl, noise_ctrl,
     COLOR_CTRL, 'Control (single fiber)'),
    (axes[1], mu_coax, std_coax, y_coax, noise_coax,
     COLOR_COAX, 'Coaxial (hollow fiber)'),
]:
    # GPR curve + bands
    ax.plot(X_pred_kpa, mu, color=color, lw=1.8, label='GPR mean')
    ax.fill_between(X_pred_kpa,
                    np.clip(mu - std, 0, None), mu + std,
                    alpha=0.35, color=color, label='±1σ (68%)')
    ax.fill_between(X_pred_kpa,
                    np.clip(mu - 2*std, 0, None), mu + 2*std,
                    alpha=0.12, color=color, label='±2σ (95%)')
    
    # Mark measured points with their noise level
    col_mean = 'coax_mean' if 'Coaxial' in label else 'ctrl_mean'
    sc = ax.scatter(df['pressure'], df[col_mean],
                    c=noise, cmap='YlOrRd', s=50, zorder=5,
                    vmin=0, vmax=noise.max())
    
    ax.set_xscale('log')
    ax.set_ylabel('Sensitivity (kPa⁻¹)', fontsize=10)
    ax.set_title(f'{label} — GPR uncertainty bands', fontsize=10)
    ax.legend(fontsize=8, loc='upper right')
    ax.grid(True, alpha=0.2)

cb = fig.colorbar(sc, ax=axes, fraction=0.02, pad=0.01)
cb.set_label('Measurement SD (colour = noise level)', fontsize=9)
axes[1].set_xlabel('Pressure (kPa)', fontsize=11)

plt.suptitle('GPR Uncertainty — darker dots = noisier measurements', fontsize=11)
plt.savefig('outputs/01_gpr_uncertainty.png', dpi=200, bbox_inches='tight')
plt.show()
print("Saved: outputs/01_gpr_uncertainty.png")
````

## 8. Query function — predict sensitivity at any pressure

````python
# ── Prediction function for deployment in the Streamlit app ───────────────
def predict_sensitivity(pressure_kpa, sensor_type='both'):
    """
    Predict sensitivity at any pressure value using the fitted GPR models.
    
    Parameters
    ----------
    pressure_kpa : float or array-like
        Pressure in kPa (can be a single value or array)
    sensor_type  : 'coaxial', 'control', or 'both'
    
    Returns
    -------
    dict with keys: pressure, coax_mean, coax_std, ctrl_mean, ctrl_std
    """
    p = np.atleast_1d(pressure_kpa)
    X_q = np.log10(p).reshape(-1, 1)
    
    results = {'pressure_kpa': p}
    
    if sensor_type in ('coaxial', 'both'):
        mu, std = gpr_coax.predict(X_q, return_std=True)
        results['coax_mean'] = np.clip(mu, 0, None)
        results['coax_std']  = std
        results['coax_ci95'] = 1.96 * std
    
    if sensor_type in ('control', 'both'):
        mu, std = gpr_ctrl.predict(X_q, return_std=True)
        results['ctrl_mean'] = np.clip(mu, 0, None)
        results['ctrl_std']  = std
        results['ctrl_ci95'] = 1.96 * std
    
    return results


# ── Demo: predict at pressures NOT in your training data ─────────────────
query_pressures = [0.3, 0.8, 1.5, 6, 12, 40, 80]
results = predict_sensitivity(query_pressures)

print("Predictions at pressures NOT in your training data:")
print(f"{'P (kPa)':>10}  {'Coaxial mean':>14}  {'±95% CI':>10}  {'Control mean':>14}  {'±95% CI':>10}")
print("-" * 65)
for i, p in enumerate(query_pressures):
    print(f"{p:>10.1f}  "
          f"{results['coax_mean'][i]:>14.4f}  "
          f"{results['coax_ci95'][i]:>10.4f}  "
          f"{results['ctrl_mean'][i]:>14.4f}  "
          f"{results['ctrl_ci95'][i]:>10.4f}")

# ── Recommend best sensor for a given pressure range ──────────────────────
def recommend_sensor(p_min, p_max, n_points=50):
    """Return which sensor type has higher mean sensitivity in a pressure range."""
    p_range = np.logspace(np.log10(p_min), np.log10(p_max), n_points)
    r = predict_sensitivity(p_range)
    coax_avg = r['coax_mean'].mean()
    ctrl_avg = r['ctrl_mean'].mean()
    winner = 'Coaxial hollow fiber' if coax_avg > ctrl_avg else 'Control single fiber'
    advantage = abs(coax_avg - ctrl_avg) / min(coax_avg, ctrl_avg) * 100
    print(f"\nFor {p_min}–{p_max} kPa:")
    print(f"  Recommended: {winner}")
    print(f"  Avg sensitivity advantage: {advantage:.1f}%")
    return winner

recommend_sensor(0.2, 3)
recommend_sensor(3, 50)
recommend_sensor(0.2, 300)
````

## 9. Results summary

````python
# ── Print complete results table ──────────────────────────────────────────
from scipy import stats as sp_stats

# Peak sensitivity with CI
n_rep = 3
t_c   = sp_stats.t.ppf(0.975, df=n_rep-1)

coax_peak_idx = df['coax_mean'].idxmax()
ctrl_peak_idx = df['ctrl_mean'].idxmax()

coax_peak_p   = df.loc[coax_peak_idx, 'pressure']
coax_peak_s   = df.loc[coax_peak_idx, 'coax_mean']
coax_peak_ci  = t_c * df.loc[coax_peak_idx, 'coax_sd'] / np.sqrt(n_rep)

ctrl_peak_p   = df.loc[ctrl_peak_idx, 'pressure']
ctrl_peak_s   = df.loc[ctrl_peak_idx, 'ctrl_mean']
ctrl_peak_ci  = t_c * df.loc[ctrl_peak_idx, 'ctrl_sd'] / np.sqrt(n_rep)

improvement = (coax_peak_s - ctrl_peak_s) / ctrl_peak_s * 100

print("=" * 60)
print("RESULTS SUMMARY")
print("=" * 60)
print(f"\nCoaxial hollow fiber sensor:")
print(f"  Peak sensitivity : {coax_peak_s:.4f} kPa⁻¹  @ {coax_peak_p} kPa")
print(f"  95% CI           : [{coax_peak_s-coax_peak_ci:.4f}, {coax_peak_s+coax_peak_ci:.4f}]")
print(f"  GPR LOO R²       : {r2_coax:.4f}")
print(f"  GPR LOO MAE      : {mae_coax:.4f} kPa⁻¹")

print(f"\nControl single fiber sensor:")
print(f"  Peak sensitivity : {ctrl_peak_s:.4f} kPa⁻¹  @ {ctrl_peak_p} kPa")
print(f"  95% CI           : [{ctrl_peak_s-ctrl_peak_ci:.4f}, {ctrl_peak_s+ctrl_peak_ci:.4f}]")
print(f"  GPR LOO R²       : {r2_ctrl:.4f}")
print(f"  GPR LOO MAE      : {mae_ctrl:.4f} kPa⁻¹")

print(f"\nImprovement (coaxial over control): {improvement:.1f}%")
print(f"Crossover pressure               : ~{cross_kpa:.1f} kPa")

print("\n" + "=" * 60)
print("FILES SAVED")
print("=" * 60)
print("  outputs/01_gpr_fit.png         — main GPR curves + CI")
print("  outputs/01_gpr_loo_cv.png      — predicted vs measured")
print("  outputs/01_gpr_uncertainty.png — uncertainty heatmap")
````

### Key concepts demonstrated (for your GitHub README)
- **Gaussian Process Regression** with Matérn kernel — correct choice for small scientific datasets  
- **Per-point noise** (`alpha=sd²`) — incorporates your experimental SD directly into the model  
- **Leave-One-Out CV** — honest performance estimate for n=33  
- **Uncertainty quantification** — posterior std gives calibrated 95% CI automatically  
- **Physics-aware preprocessing** — log10 pressure transform respects the multi-decade x-axis  
- **Deployment-ready** — `predict_sensitivity()` function slots directly into the Streamlit app
